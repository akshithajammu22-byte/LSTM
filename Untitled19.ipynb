```python
commands = ["on", "off", "go", "stop"]

for command in commands:
    folder = os.path.join(DATASET_PATH, command)
    
    if os.path.exists(folder):
        wav_files = [
            f for f in os.listdir(folder)
            if f.lower().endswith(".wav")
        ]
        
        print(f"{command}: {len(wav_files)} audio files")
    else:
        print(f"{command}: NOT FOUND")
```

```python
from IPython.display import Audio, display

command = "on"

folder = os.path.join(DATASET_PATH, command)

wav_files = [
    f for f in os.listdir(folder)
    if f.lower().endswith(".wav")
]

audio_file = os.path.join(folder, wav_files[0])

print("Audio file:")
print(audio_file)

display(Audio(audio_file))
```

```python
audio, sample_rate = librosa.load(
    audio_file,
    sr=16000
)

print("Sample rate:", sample_rate)
print("Number of samples:", len(audio))
print("Duration:", len(audio) / sample_rate, "seconds")
```

```python
plt.figure(figsize=(12, 4))

librosa.display.waveshow(
    audio,
    sr=sample_rate
)

plt.title("Speech Waveform - ON")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")

plt.show()
```

```python
N_MFCC = 40
MAX_LEN = 32

mfcc = librosa.feature.mfcc(
    y=audio,
    sr=sample_rate,
    n_mfcc=N_MFCC
)

print("MFCC shape:", mfcc.shape)
```

```python
plt.figure(figsize=(10, 4))

librosa.display.specshow(
    mfcc,
    x_axis="time",
    sr=sample_rate
)

plt.colorbar()

plt.title("MFCC Features")
plt.xlabel("Time")
plt.ylabel("MFCC Coefficients")

plt.show()
```

```python
def extract_mfcc(file_path):
    
    audio, sr = librosa.load(
        file_path,
        sr=16000
    )
    
    mfcc = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=N_MFCC
    )
    
    # Padding if audio has fewer frames
    if mfcc.shape[1] < MAX_LEN:
        
        pad_width = MAX_LEN - mfcc.shape[1]
        
        mfcc = np.pad(
            mfcc,
            ((0, 0), (0, pad_width)),
            mode="constant"
        )
    
    # Truncate if audio has more frames
    else:
        mfcc = mfcc[:, :MAX_LEN]
    
    # Convert:
    # (40, 32) → (32, 40)
    return mfcc.T
```

```python
features = extract_mfcc(audio_file)

print("Feature shape:", features.shape)
```

```python
X = []
y = []

for command in commands:
    
    folder = os.path.join(
        DATASET_PATH,
        command
    )
    
    wav_files = [
        f for f in os.listdir(folder)
        if f.lower().endswith(".wav")
    ]
    
    print(
        f"Processing {command}: "
        f"{len(wav_files)} files"
    )
    
    for file in wav_files:
        
        file_path = os.path.join(
            folder,
            file
        )
        
        try:
            features = extract_mfcc(file_path)
            
            X.append(features)
            y.append(command)
            
        except Exception as e:
            print("Error:", file_path)
            print(e)

print("\nDataset processing completed!")
print("Total samples:", len(X))
```

```python
X = np.array(X)
y = np.array(y)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
unique, counts = np.unique(
    y,
    return_counts=True
)

for label, count in zip(unique, counts):
    print(label, ":", count)
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

classes = label_encoder.classes_

print("Classes:")
print(classes)

print("\nEncoded values:")
print(np.unique(y_encoded))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)
```

```python
model = Sequential([
    
    LSTM(
        128,
        input_shape=(MAX_LEN, N_MFCC),
        return_sequences=True
    ),
    
    Dropout(0.3),
    
    LSTM(64),
    
    Dropout(0.3),
    
    Dense(
        64,
        activation="relu"
    ),
    
    Dropout(0.3),
    
    Dense(
        len(classes),
        activation="softmax"
    )
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled!")
```

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)
```

```python
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=30,
    batch_size=32,
    callbacks=[early_stopping]
)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("LSTM Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()

plt.show()
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print(
    "Test Accuracy:",
    f"{test_accuracy * 100:.2f}%"
)
```

```python
y_pred_probability = model.predict(
    X_test
)

y_pred = np.argmax(
    y_pred_probability,
    axis=1
)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=classes
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=classes,
    yticklabels=classes
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Wake-Word LSTM Confusion Matrix")

plt.show()
```

```python
def predict_audio(file_path):
    
    features = extract_mfcc(file_path)
    
    # Add batch dimension
    features = np.expand_dims(
        features,
        axis=0
    )
    
    probabilities = model.predict(
        features,
        verbose=0
    )[0]
    
    predicted_index = np.argmax(
        probabilities
    )
    
    predicted_class = classes[
        predicted_index
    ]
    
    confidence = probabilities[
        predicted_index
    ]
    
    return predicted_class, confidence
```

```python
result, confidence = predict_audio(
    audio_file
)

print("Prediction:", result)
print(
    "Confidence:",
    f"{confidence * 100:.2f}%"
)
```

```python
def wake_word_prediction(
    file_path,
    threshold=0.80
):
    
    predicted_class, confidence = predict_audio(
        file_path
    )
    
    if confidence >= threshold:
        return predicted_class, confidence
    
    return "unknown", confidence
```

```python
def activate_device(command):
    
    if command == "on":
        print("🟢 SMART DEVICE: ON")
        
    elif command == "off":
        print("🔴 SMART DEVICE: OFF")
        
    elif command == "go":
        print("▶ SMART DEVICE STARTED")
        
    elif command == "stop":
        print("⏹ SMART DEVICE STOPPED")
        
    else:
        print("⚪ No activation")
```

```python
activate_device("on")
```

```python
def process_audio(file_path):
    
    command, confidence = wake_word_prediction(
        file_path,
        threshold=0.80
    )
    
    print(
        "Detected:",
        command
    )
    
    print(
        "Confidence:",
        f"{confidence * 100:.2f}%"
    )
    
    activate_device(command)
```

```python
process_audio(audio_file)
```
